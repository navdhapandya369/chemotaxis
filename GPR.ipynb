```python
import numpy as np
import pandas as pd
import pde
from scipy.stats.qmc import LatinHypercube
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel
import matplotlib.pyplot as plt

# ============================================================
# 1. FIXED CONSTANTS
# ============================================================
K_eq  = 100.0
delta = 0.8
D_E   = 40.0
D_EL  = delta * D_E       # 32.0
D_S   = 10.0 * D_E        # 400.0
D_P   = 10.0 * D_E        # 400.0
E0    = 1.0
L     = 20.0

print("Fixed constants:")
print(f"  K_eq  = {K_eq}")
print(f"  delta = {delta}")
print(f"  D_E   = {D_E}  (μm)²/s")
print(f"  D_EL  = {D_EL} (μm)²/s")
print(f"  D_S   = {D_S}  (μm)²/s")
print(f"  D_P   = {D_P}  (μm)²/s")
print(f"  E0    = {E0}   mM")
print(f"  L     = {L}    μm")

# ============================================================
# 2. VARIABLE RANGES
# ============================================================
# k2 = kcat = kcat0/(1+alpha*S) — not a free variable
# k-2 = k1*kcat0/(km1*Keq*(1+alpha*S)) — from K_eq constraint
# So free variables are: k1, km1, S0, kcat0, alpha

ranges = {
    "k1":    [10,   500],    # mM⁻¹s⁻¹  forward binding
    "km1":   [10,   500],    # s⁻¹       unbinding
    "S0":    [0.1,  10.0],   # mM        reservoir concentration
    "kcat0": [100,  5000],   # s⁻¹       base catalytic rate
    "alpha": [0.01, 5.0],    # mM⁻¹      substrate inhibition
}

var_names = list(ranges.keys())   # ["k1","km1","S0","kcat0","alpha"]
bounds    = np.array(list(ranges.values()))   # shape (5,2)
n_vars    = len(ranges)           # 5

print("\nBO Variable Ranges:")
print("─" * 45)
for var, (lo, hi) in ranges.items():
    print(f"  {var:8s}: [{lo:8.3f},  {hi:8.3f}]")

# ============================================================
# 3. LATIN HYPERCUBE SAMPLING
# ============================================================
n_samples = 30

sampler      = LatinHypercube(d=n_vars, seed=42)
unit_samples = sampler.random(n=n_samples)   # shape (30,5), values in [0,1]

# scale to actual ranges
lo     = bounds[:, 0]
hi     = bounds[:, 1]
scaled = lo + unit_samples * (hi - lo)       # shape (30,5)

df_lhs = pd.DataFrame(scaled, columns=var_names)

print(f"\nLHS: {n_samples} samples generated")
print("First 5:")
print(df_lhs.round(3).head())

# ============================================================
# 4. SIMULATION FUNCTION
# ============================================================
def run_simulation(k1, km1, S0, kcat0, alpha):
    """
    Runs py-pde simulation with:
      k2  = kcat0/(1+alpha*S)        — varies spatially
      km2 = k1*kcat0/(km1*Keq*(1+alpha*S)) — varies spatially

    Returns:
      x_max    — location of E+EL peak (None if no interior peak)
      strength — max(E+EL) - min(E+EL)
      label    — 1 if interior peak exists, 0 if not
    """

    grid = pde.CartesianGrid([[0, L]], [200], periodic=False)

    eq = pde.PDE(
        {
            # k2  = kcat0/(1+alpha*S)
            # km2 = k1*kcat0/(km1*Keq*(1+alpha*S))
            # both depend on local S(x,t)

            "E":  """D_E*laplace(E) - k1*E*S + km1*EL - (k1*kcat0/(km1*Keq*(1+alpha*S)))*P*E + (kcat0/(1+alpha*S))*EL""",

            "EL": """D_EL*laplace(EL) + k1*E*S + (k1*kcat0/(km1*Keq*(1+alpha*S)))*P*E - km1*EL - (kcat0/(1+alpha*S))*EL""",

            "S":  """D_S*laplace(S) - k1*E*S + km1*EL""",

            "P":  """D_P*laplace(P) + (kcat0/(1+alpha*S))*EL - (k1*kcat0/(km1*Keq*(1+alpha*S)))*P*E""",
        },
        consts={
            "D_E":   D_E,
            "D_EL":  D_EL,
            "D_S":   D_S,
            "D_P":   D_P,
            "k1":    k1,
            "km1":   km1,
            "kcat0": kcat0,
            "alpha": alpha,
            "Keq":   K_eq,
        },
        bc_ops={
            "E:*":  "neumann",
            "EL:*": "neumann",
            "S:*":  ({"value": 0.9*S0}, {"value": 0.1*S0}),
            "P:*":  ({"value": 0.1*S0}, {"value": 0.9*S0}),
        },
    )

    # initial conditions
    E_f   = pde.ScalarField(grid, E0,  label="E")
    EL_f  = pde.ScalarField(grid, 0.0, label="EL")
    S_f   = pde.ScalarField(grid, 0.0, label="S")
    P_f   = pde.ScalarField(grid, 0.0, label="P")
    state = pde.FieldCollection([E_f, EL_f, S_f, P_f])

    # solve
    try:
        result = eq.solve(state, t_range=200, dt=None)
    except Exception as e:
        print(f"  Simulation failed: {e}")
        return None, 0.0, 0

    # extract E+EL profile
    x_coords = grid.axes_coords[0]
    E_total  = result[0].data + result[1].data

    # check for unphysical results
    if np.any(np.isnan(E_total)) or np.any(E_total < 0):
        print(f"  Unphysical result — skipping")
        return None, 0.0, 0

    # find interior extrema
    gradient     = np.gradient(E_total, x_coords)
    sign_changes = np.where(np.diff(np.sign(gradient)))[0]
    interior     = [i for i in sign_changes
                    if i > 2 and i < len(x_coords) - 3]

    if len(interior) > 0:
        x_max    = float(x_coords[np.argmax(E_total)])
        strength = float(E_total.max() - E_total.min())
        label    = 1
    else:
        x_max    = None
        strength = 0.0
        label    = 0

    return x_max, strength, label

# ============================================================
# 5. GENERATE DUMMY OUTPUTS — run simulation at LHS points
# ============================================================
results = []

print(f"\nRunning {n_samples} LHS simulations...")
print("─" * 60)

for i, row in df_lhs.iterrows():
    print(f"Sample {i+1:2d}/{n_samples}: ", end="", flush=True)

    x_max, strength, label = run_simulation(
        row["k1"], row["km1"],
        row["S0"], row["kcat0"], row["alpha"]
    )

    results.append({
        "k1":       row["k1"],
        "km1":      row["km1"],
        "S0":       row["S0"],
        "kcat0":    row["kcat0"],
        "alpha":    row["alpha"],
        "x_max":    x_max,
        "strength": strength,
        "label":    label,
    })

    print(f"x_max={x_max}, "
          f"strength={strength:.5f}, "
          f"label={label}")

df_results = pd.DataFrame(results)

print("\nFull dummy output table:")
print("─" * 80)
print(df_results.round(3).to_string())

# summary
n_valid   = df_results["label"].sum()
n_invalid = n_samples - n_valid
print(f"\nValid (label=1):   {n_valid}/{n_samples}")
print(f"Invalid (label=0): {n_invalid}/{n_samples}")

# ============================================================
# 6. FIT GPR — use only valid samples (label=1)
# ============================================================
df_valid = df_results[
    (df_results["label"] == 1) &
    (df_results["x_max"].notna())
].copy()

print(f"\n{len(df_valid)} valid samples used for GPR fitting")

# training data
X_train    = df_valid[var_names].values     # shape (n_valid, 5)
y_xmax     = df_valid["x_max"].values       # peak location
y_strength = df_valid["strength"].values    # peak strength

# normalise inputs — important for GPR
X_mean = X_train.mean(axis=0)
X_std  = X_train.std(axis=0)
X_norm = (X_train - X_mean) / X_std

print("\nInput normalisation:")
for i, name in enumerate(var_names):
    print(f"  {name:8s}: mean={X_mean[i]:.3f}, std={X_std[i]:.3f}")

# ============================================================
# 7. DEFINE KERNEL (PRIOR)
# ============================================================
# This is where the prior is chosen — as your professor said,
# prior comes after fitting GPR to LHS data
# RBF kernel = smooth function assumption
# One length scale per variable = each parameter treated independently

kernel = ConstantKernel(
    constant_value=1.0,
    constant_value_bounds=(1e-3, 1e3)
) * RBF(
    length_scale=np.ones(n_vars),
    length_scale_bounds=(1e-2, 1e2)
)

print(f"\nKernel (prior) chosen: ConstantKernel × RBF")
print(f"  One length scale per variable: {n_vars} length scales")

# ============================================================
# 8. FIT GPR FOR x_max
# ============================================================
gpr_xmax = GaussianProcessRegressor(
    kernel=kernel,
    n_restarts_optimizer=10,
    normalize_y=True,
    alpha=1e-6,
)
gpr_xmax.fit(X_norm, y_xmax)

print(f"\nGPR for x_max fitted:")
print(f"  Kernel: {gpr_xmax.kernel_}")
print(f"  Log marginal likelihood: "
      f"{gpr_xmax.log_marginal_likelihood_value_:.3f}")

# ============================================================
# 9. FIT GPR FOR STRENGTH
# ============================================================
gpr_strength = GaussianProcessRegressor(
    kernel=kernel,
    n_restarts_optimizer=10,
    normalize_y=True,
    alpha=1e-6,
)
gpr_strength.fit(X_norm, y_strength)

print(f"\nGPR for strength fitted:")
print(f"  Kernel: {gpr_strength.kernel_}")
print(f"  Log marginal likelihood: "
      f"{gpr_strength.log_marginal_likelihood_value_:.3f}")

# ============================================================
# 10. CHECK GPR QUALITY on training data
# ============================================================
y_pred_xmax, sig_xmax = gpr_xmax.predict(
    X_norm, return_std=True
)
y_pred_str, sig_str = gpr_strength.predict(
    X_norm, return_std=True
)

print("\nGPR predictions vs actual (training data):")
print("─" * 65)
print(f"{'#':>3} {'Actual x_max':>14} {'Predicted':>12} "
      f"{'σ':>8} {'Error':>10}")
print("─" * 65)
for i in range(len(df_valid)):
    error = abs(y_xmax[i] - y_pred_xmax[i])
    print(f"{i+1:>3} {y_xmax[i]:>14.3f} "
          f"{y_pred_xmax[i]:>12.3f} "
          f"{sig_xmax[i]:>8.3f} "
          f"{error:>10.3f}")

# ============================================================
# 11. PREDICT AT NEW UNTRIED PARAMETER SET
# ============================================================
new_params = np.array([[
    200,    # k1
    150,    # km1
    2.0,    # S0
    800,    # kcat0
    1.0,    # alpha
]])

new_norm = (new_params - X_mean) / X_std

mu_xmax,     sig_xmax_new = gpr_xmax.predict(
    new_norm, return_std=True
)
mu_strength, sig_str_new  = gpr_strength.predict(
    new_norm, return_std=True
)

print(f"\nPrediction at new (untried) parameter set:")
print(f"  k1    = {new_params[0,0]}")
print(f"  km1   = {new_params[0,1]}")
print(f"  S0    = {new_params[0,2]}")
print(f"  kcat0 = {new_params[0,3]}")
print(f"  alpha = {new_params[0,4]}")
print(f"  → Predicted x_max:    "
      f"{mu_xmax[0]:.3f} ± {sig_xmax_new[0]:.3f} μm")
print(f"  → Predicted strength: "
      f"{mu_strength[0]:.5f} ± {sig_str_new[0]:.5f}")

# ============================================================
# 12. PLOTS
# ============================================================
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# plot 1: GPR fit for x_max
axes[0].errorbar(
    y_xmax, y_pred_xmax,
    yerr=sig_xmax,
    fmt='o', color='blue', alpha=0.7,
    capsize=4, label='GPR ± σ'
)
axes[0].plot(
    [y_xmax.min(), y_xmax.max()],
    [y_xmax.min(), y_xmax.max()],
    'r--', linewidth=2, label='Perfect fit'
)
axes[0].set_xlabel("Actual x_max (μm)")
axes[0].set_ylabel("Predicted x_max (μm)")
axes[0].set_title("GPR Fit: Peak Location (x_max)")
axes[0].legend()
axes[0].grid(alpha=0.3)

# plot 2: GPR fit for strength
axes[1].errorbar(
    y_strength, y_pred_str,
    yerr=sig_str,
    fmt='o', color='green', alpha=0.7,
    capsize=4, label='GPR ± σ'
)
axes[1].plot(
    [y_strength.min(), y_strength.max()],
    [y_strength.min(), y_strength.max()],
    'r--', linewidth=2, label='Perfect fit'
)
axes[1].set_xlabel("Actual strength")
axes[1].set_ylabel("Predicted strength")
axes[1].set_title("GPR Fit: Peak Strength")
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

# ============================================================
# 13. SAVE RESULTS
# ============================================================
df_results.to_csv("lhs_dummy_outputs.csv", index=False)
print("\nResults saved to lhs_dummy_outputs.csv")
print("\nGPR fitted and ready for BO loop.")
print("Next step: use gpr_xmax and gpr_strength")
print("in the acquisition function to run BO.")
```
